```python

from google.colab import drive
drive.mount('/content/drive')

import os

PROJECT_PATH = "/content/drive/MyDrive/Colab Notebooks/IDSE_PROJECT"

DATA_PATH = os.path.join(PROJECT_PATH, "processed_data")
MODEL_PATH = os.path.join(PROJECT_PATH, "models")
RESULT_PATH = os.path.join(PROJECT_PATH, "results")

os.makedirs(DATA_PATH, exist_ok=True)
os.makedirs(MODEL_PATH, exist_ok=True)
os.makedirs(RESULT_PATH, exist_ok=True)

print("Project folder connected successfully")
```

```python

import pandas as pd
import numpy as np

from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score

X_train = pd.read_csv(os.path.join(DATA_PATH, "X_train.csv"))
X_test = pd.read_csv(os.path.join(DATA_PATH, "X_test.csv"))

y_train = pd.read_csv(os.path.join(DATA_PATH, "y_train.csv")).values.ravel()
y_test = pd.read_csv(os.path.join(DATA_PATH, "y_test.csv")).values.ravel()

print("Training shape:", X_train.shape)
print("Testing shape:", X_test.shape)
```

```python
from sklearn.dummy import DummyClassifier

baseline = DummyClassifier(
    strategy="most_frequent"
)

baseline.fit(X_train, y_train)

baseline_score = baseline.score(
    X_test, y_test
)

print("Baseline Accuracy:", baseline_score)
```

```python

baseline_pred = baseline.predict(X_test)

baseline_results = pd.DataFrame({
    "Model": ["Baseline"],
    "Accuracy": [accuracy_score(y_test, baseline_pred)]
})

baseline_results.to_csv(
    os.path.join(RESULT_PATH, "baseline_results.csv"),
    index=False
)
```
