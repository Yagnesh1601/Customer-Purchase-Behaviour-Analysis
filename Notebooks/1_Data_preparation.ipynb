```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Load the dataset using the correct project path variable
file_path = os.path.join(PROJECT_PATH, "Online Retail.xlsx")
df = pd.read_excel(file_path)

print("Dataset loaded successfully!")
print("Number of rows and columns:", df.shape)

df.head()
```

```python
print("Column names:")
print(df.columns)

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nStatistical Summary:")
df.describe()
```

```python
print("Duplicate rows:", df.duplicated().sum())

print("Cancelled invoices:",
      df["InvoiceNo"].astype(str).str.startswith("C").sum())

print("Negative quantities:",
      (df["Quantity"] < 0).sum())

print("Invalid prices:",
      (df["UnitPrice"] <= 0).sum())

print("Missing Customer IDs:",
      df["CustomerID"].isnull().sum())
```

```python
sales_df = df.copy()

sales_df = sales_df.dropna(subset=["CustomerID"])

sales_df = sales_df[
    (sales_df["Quantity"] > 0) &
    (sales_df["UnitPrice"] > 0)
]

sales_df = sales_df[
    ~sales_df["InvoiceNo"].astype(str).str.startswith("C")
]

sales_df["Revenue"] = (
    sales_df["Quantity"] * sales_df["UnitPrice"]
)

print("Cleaned dataset shape:", sales_df.shape)

sales_df.head()
```

```python
sales_df.to_csv(
    os.path.join(DATA_PATH, "cleaned_online_retail.csv"),
    index=False
)

print("Cleaned dataset saved successfully!")
```
